```python
'''
8. Write a program to display the following pattern: 
*****
*   *
*   *
*   *
*****
'''
def square(n):
    for i in range(n):
        for j in range(n):
            if i==0 or i==n-1 or j==0 or j==n-1:
                print("*", end="")
            else:
                print(" ", end="")
        print()
square(5)
```
