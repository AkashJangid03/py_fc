```python
# 12. Write a program to calculate the sum of the following series Sum=1+ 1/2! + 1/3! + 1/4! ... + 1/n!
def series_sum(n):
    total = 1
    fact = 1
    for i in range(2, n + 1):
        fact = fact * i
        total = total + 1 / fact
    return total
n = int(input("Enter n: "))
print("Sum =", series_sum(n))
```
