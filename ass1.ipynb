```python
#1. Write a program to determine all the Pythagorean triplets in the range of 100 to 1000. A Pythagorean triplet is a set of three integers I, j, k such that (i^2 + j^2 = k^2)  
for i in range(100, 1001):
    for j in range(i, 1001):
        for k in range(j, 1001):
            if i * i + j * j == k * k:
                print(i, j, k)
```

```python
def add1(n):
    res = 0
    place = 1

    while n > 0:
        digit = n % 10
        digit = (digit + 1) % 10
        res = res + digit * place
        place = place * 10
        n //= 10
    print(res)
n = int(input("Enter a number: "))
add1(n)
```

```python
#4. Write a Python program to calculate the sum of the first 10 prime numbers. 
count = 0
num = 2
total = 0
while count < 10:
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        total = total + num
        count = count + 1
    num = num + 1
print("Sum of first 10 prime numbers =", total)
```

```python
# 6. Write a python program to find all the Armstrong numbers within a given range.
def armstrong_number(start, end):
    for i in range(start, end + 1):
        order = len(str(i))
        temp = i
        digit_sum = 0
        while temp > 0:
            digit = temp % 10
            digit_sum += digit ** order
            temp //= 10
        if i == digit_sum:
            print(i, end=" ")
start = int(input("starting number: "))
end = int(input("ending number: "))
armstrong_number(start, end)
```

```python
# 7. Write a python program to find all the Strong numbers within a given range.
import math
def strongnum(start, end): 
    for i in range(start, end+1): 
        temp = i
        total = 0
        while temp > 0:
            digit = temp %10
            total += math.factorial(digit)
            temp //=10
            if i == total:
                print(i, end=" ")
start = int(input("start num: "))
end = int(input("end num: "))
strongnum(start, end)
```

```python
'''
8. Write a program to display the following pattern: 
*****
*   *
*   *
*   *
*****
'''
def square(n):
    for i in range(n):
        for j in range(n):
            if i==0 or i==n-1 or j==0 or j==n-1:
                print("*", end="")
            else:
                print(" ", end="")
        print()
square(5)
```

```python
# 9. Write a program to print all integers between 1 and n that are not divisible by either 2 or 3. The program should also count and display the number of such integers. 
def pnum(n):
    count=0
    for i in range(1, n+1):
        if i%2 != 0 and i%3!=0:
            print(i, end=" ")
            count+=1
    print("count: ", count)
n=int(input("Enter the number: "))
pnum(n)
```

```python
'''
10. Write a program to display the following pattern: 
*****
*****
**@**
*****
***** 
'''
def pattern(n):
    for i in range(n):
        for j in range(n):
            if i == n // 2 and j == n // 2:
                print("@", end="")
            else:
                print("*", end="")
        print()

pattern(5)
```

```python
## 11. pascal tree
def pascal(n):
    for i in range(n):
        num = 1
        for j in range(i + 1):
            print(num, end=" ")
            num = num * (i - j) // (j + 1)
        print()
n = int(input("Enter number of rows: "))
pascal(n)
```

```python
# 12. Write a program to calculate the sum of the following series Sum=1+ 1/2! + 1/3! + 1/4! ... + 1/n!
def series_sum(n):
    total = 1
    fact = 1
    for i in range(2, n + 1):
        fact = fact * i
        total = total + 1 / fact
    return total
n = int(input("Enter n: "))
print("Sum =", series_sum(n))
```

```python
def tables(start, end):
    for i in range(start, end + 1):
        if i % 2 != 0:
            print("\nTable of", i)
            for j in range(1, 11):
                print(i, "x", j, "=", i * j)
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))
tables(start, end)
```

```python
def triangle(n):
    for i in range(1, n + 1):
        print(" " * (n - i), end="")
        for j in range(i):
            print("*", end=" ")
        print()
n = int(input("Enter number of rows: "))
triangle(n)
```
