```python
#4. Write a Python program to calculate the sum of the first 10 prime numbers. 
count = 0
num = 2
total = 0
while count < 10:
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        total = total + num
        count = count + 1
    num = num + 1
print("Sum of first 10 prime numbers =", total)
```
